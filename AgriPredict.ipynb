```python
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder
from sklearn.model_selection import KFold, cross_validate
from sklearn.pipeline import make_pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.dummy import DummyClassifier
```

```python
preprocessor = make_column_transformer(
    (StandardScaler(), ['pH_sol', 'azote_N_kg_ha', 'phosphore_P_kg_ha', 'potassium_K_kg_ha',
       'pluviometrie_mm', 'temperature_moy_C', 'humidite_air_pct']),
    (OrdinalEncoder(categories = ord_), ['niveau_ensoleillement', 'qualite_drainage', 'niveau_fertilite_sol']),
    (OneHotEncoder(drop = "first", handle_unknown = "ignore"), ['type_sol', 'region', 'methode_irrigation'])
)
```

```python
model = make_pipeline(preprocessor, KNeighborsClassifier(n_neighbors = 5))
```

```python
X = data
```

```python
kf = KFold(n_splits = 5, shuffle = True, random_state = 42)
score = cross_validate(model, X, target, cv =kf, scoring = "accuracy" )
score
```

```python
score = score["test_score"].mean()
```

```python
print(score)
```
